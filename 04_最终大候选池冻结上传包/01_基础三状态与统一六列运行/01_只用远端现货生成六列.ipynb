默认读取此前约定的 `COMPANY_SPOT_PATH` 或远端现货通配路径。若路径不存在，Notebook 会直接报错，不会偷偷使用本地数据。

```python
from pathlib import Path
import sys

def find_package_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / 'src' / 'generate_compact_output.py').is_file():
            return candidate
    raise FileNotFoundError('没有找到 06_最终上传包 根目录。请确认 Notebook 与 src、packages 位于同一目录。')

PACKAGE_ROOT = find_package_root(Path.cwd().resolve())
sys.path.insert(0, str(PACKAGE_ROOT / 'src'))
from generate_compact_output import generate_compact_output

record = generate_compact_output()
record
```
